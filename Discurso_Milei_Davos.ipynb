```python
from collections import Counter
import pandas as pd
import spacy

doc = nlp(texto)

# Genero una lista de lemas ignorando stop words y puntuación
lemas = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]

# Cuento los lemas
conteo = Counter(lemas)

# El conteo pasa a ser un DataFrame
df_conteo = pd.DataFrame(conteo.items(), columns=['Lema', 'Frecuencia'])

# Ordeno de mayor a menor frecuencia
df_conteo = df_conteo.sort_values(by='Frecuencia', ascending=False).reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt

# Selecciono las 10 palabras más frecuentes
top_10 = df_conteo.head(10)

# Creo el gráfico de barras
plt.figure(figsize=(10, 6))
plt.bar(top_10['Lema'], top_10['Frecuencia'], color='skyblue')

plt.title('Top 10 palabras más usadas', fontsize=16)
plt.xlabel('Palabras', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.yticks(fontsize=10)

plt.show()
```
